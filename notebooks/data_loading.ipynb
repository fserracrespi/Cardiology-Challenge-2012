# Data Download: PhysioNet/CinC Challenge 2012

## Description

This notebook retrieves the **PhysioNet/CinC Challenge 2012** dataset [1], which consists of irregularly sampled multivariate time series from patients admitted to intensive care units (ICU). Each record covers the first 48 hours of a patient's stay.

Two public subsets are used:

| Subset | File        | Use in this work          | Target directory             |
|--------|-------------|---------------------------|------------------------------|
| Set A  | `set-a.zip` | Training and validation   | `data/raw/Train_Val/set-a/`  |
| Set B  | `set-b.zip` | Evaluation (test)         | `data/raw/Test/set-b/`       |

**Reference**

[1] Silva, I., Moody, G., Scott, D. J., Celi, L. A., & Mark, R. G. (2012). *Predicting in-hospital mortality of ICU patients: The PhysioNet/Computing in Cardiology Challenge 2012*. Computing in Cardiology, 39, 245–248.

## 1. Configuration

```python
import urllib.request
import zipfile
from pathlib import Path

BASE_URL = "https://physionet.org/files/challenge-2012/1.0.0"
RAW_DIR = Path.cwd().parent / "data" / "raw"
```

## 2. Download and Extraction Function

The `download_and_extract` function downloads the given compressed file and extracts it into the corresponding directory.

```python
def download_and_extract(filename: str, split_dir: str):
    """Download `filename` from PhysioNet into RAW_DIR/split_dir and extract it."""
    out_dir = RAW_DIR / split_dir
    out_dir.mkdir(parents=True, exist_ok=True)

    zip_path = out_dir / filename
    extract_dir = out_dir / zip_path.stem

    print(f"{filename}: downloading...")
    urllib.request.urlretrieve(f"{BASE_URL}/{filename}", zip_path)

    with zipfile.ZipFile(zip_path, "r") as zip_ref:
        zip_ref.extractall(extract_dir)

    print(f"{filename}: extracted to {extract_dir}")
    return extract_dir
```

## 3. Subset Download

### 3.1. Set A (training and validation)

```python
train_val_dir = download_and_extract("set-a.zip", "Train_Val")
```

### 3.2. Set B (test)

```python
test_dir = download_and_extract("set-b.zip", "Test")
```

## 4. Verification

The number of extracted records in each subset is checked. According to the challenge documentation, each subset should contain 4000 patients (one `.txt` file per patient).

```python
for name, path in [("Set A", train_val_dir), ("Set B", test_dir)]:
    n_records = sum(1 for _ in path.rglob("*.txt"))
    print(f"{name}: {n_records} records in {path}")
```
